```python

#Фабрика массивов
zeros_array = np.zeros(7)
print(zeros_array)
print(zeros_array.dtype)
print(zeros_array.itemsize)
print(zeros_array.nbytes)

zeros_array_int = np.zeros(7, dtype=np.int64)
print(zeros_array_int)
print(zeros_array_int.dtype)
print(zeros_array_int.itemsize)
print(zeros_array_int.nbytes)

zeros_array_int = np.zeros(7, dtype=int)
print(zeros_array_int)
print(zeros_array_int.dtype)

ones_array = np.ones(7)
print(ones_array)

empty_array = np.empty(7)
print(empty_array)
```

```python
import time
from tqdm.auto import tqdm
tqdm.pandas()


def get_price_group1(price):
    time.sleep(0.01)
    if price < 1000:
        return "До 1000 евро"
    return "Более 1000 евро"


df["Price_euros"].progress_apply(get_price_group1)
```

```python
demo = df[["Company", "Product", "Price_euros", "Inches"]].head(10).copy()
demo
```

```python
def get_offer_label(price, inches):
    if price < 1000 and inches <= 14:
        return "Подходит"
    return "Не подходит"
```

```python
first_row = demo.iloc[0]
first_row
```

```python
first_result = get_offer_label(first_row["Price_euros"], first_row["Inches"])
first_result
```

```python
def class_row(row):
    price = row["Price_euros"]
    inches = row["Inches"]
    return get_offer_label(price, inches)
```

```python
demo["offer_label"] = demo.apply(class_row, axis=1)
demo
```
